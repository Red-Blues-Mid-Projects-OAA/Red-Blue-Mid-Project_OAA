```python
import pandas as pd
import numpy as np
import os  
import yfinance as yf
from datetime import datetime, timedelta
```

```python
# 티커 정의

tickers = ['NVDA', 'GOOGL', 'AAPL', 'MSFT', 'AMZN', 'META', 'TSM', 'TSLA', 'AVGO', 'BRK-A']
```

```python
start_date = "2015-01-01"# 문자열 야후파이낸스에서
end_date = datetime.now().strftime("%Y-%m-%d")# 현재시간반환
                        # 문자열 포멧 시간변환
```

```python
print(f"Fetching data for: {', '.join(tickers)}") #Example: '.'.join(['ab', 'pq', 'rs']) -> 'ab.pq.rs'
print(f"Data range: {start_date} to {end_date} (Exclusive)")
```

```python
# yfinance를 통해 수정 종가 데이터 다운로드
data = yf.download(tickers, start=start_date, end=end_date, auto_adjust=True)['Close']#수정종가를 가져오기위한 방법
# ['Close']가 없다면 그날의 고점 저점 거래량 종가 가나타남
```
